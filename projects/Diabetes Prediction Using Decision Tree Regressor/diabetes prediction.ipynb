```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_diabetes
```

```python
dataset = load_diabetes()
```

```python
print(dataset['DESCR'])
```

```python
df_diabetes = pd.DataFrame(dataset.data,columns=dataset.feature_names)
```

```python
df_diabetes.head()
```

```python
x = df_diabetes
y = dataset['target']
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
plt.figsize=(15,5)
sns.heatmap(x_train.corr(),annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param = {
  'criterion':['squared_error','friedman_mse','absolute_error'],
  'splitter':['best','random'],
  'max_depth':[1,2,3,4,5,10,15,20,25],
  'max_features':['auto','sqrt','log2']
}
grid = GridSearchCV(regressor,param_grid=param,cv=5,scoring='neg_mean_squared_error')
```

```python
param
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
r2 = r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
print("mse:",mse)
print("mae:",mae)
print("r2:",r2)
```

```python
selectModel = DecisionTreeRegressor(criterion='friedman_mse',max_depth=3,max_features='log2',splitter='best')
```

```python
selectModel.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(selectModel,filled=True)
plt.show()
```
